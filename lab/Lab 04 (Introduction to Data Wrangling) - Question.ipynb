Task 1: Pandas Series 
Create Pandas Series from lists. Run the following code and observe the output.

```python
#import the modules for pandas and numpy

import pandas as pd
import numpy as np

# create two lists
emp_name_list = ["Alice", "Bob", "Cathy"]
emp_height_list = [168, 170, 155]

# create series from lists
emp_name_series = pd.Series(emp_name_list, name="Employee_Name")
emp_height_series = pd.Series(emp_height_list, name="Employee_Height")

print("\nPrint Series")
print(emp_name_series, "\n")
print(emp_height_series)

print("\nAccess a particular element using index")
print(emp_name_series[1])
print(emp_height_series[1])

```

**Task 2: Pandas DataFrame**

Create a Pandas DataFrame. Run the following code and observe the output. Discuss the difference between Series and DataFrame.


```python
import pandas as pd
import numpy as np

# create a dictionary
emp_profile = {"Name": emp_name_series, "Height": emp_height_series}

# create a dataframe from dictionary
emp_df = pd.DataFrame(emp_profile)

print("\nPrint dataframe")
print(emp_df)

print("\nSelect a column")
print(emp_df['Name'])

print("\nSelect multiple columns")
print(emp_df[['Name','Height']])

print("\nSelect a row")
print(emp_df.iloc[2])

print("\nSelect multiple rows")
print(emp_df.iloc[[1, 2]])

print("\nOther examples")
print(emp_df["Name"][0])
print(emp_df["Height"][1])

#Sample solution: Series is a one-dimensional array whereas DataFrame consists of rows and columns. 
```

**Task 4: NumPy vs Python List**

Run the following code and observe the difference in the loop execution time between Python list and NumPy. Which structure is faster? 


```python
import pandas as pd
import numpy as np
import time

size_vec = 2000000
t1 = time.time()
X = range(size_vec)
Y = range(size_vec)
Z = [X[i]+Y[i] for i in range(len(X))]
list_time = time.time()-t1

t2 = time.time()
X2 = np.arange(size_vec)
Y2 = np.arange(size_vec)
Z2 = X2+Y2
list_time2 = time.time()-t2

print(list_time, list_time2)

#Numpy is faster than list
```
