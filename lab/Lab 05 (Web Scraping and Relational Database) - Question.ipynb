```python
#Connect to the database company.db. (If the database does not already exist, it will be created.)

import sqlite3

conn = sqlite3.connect("company.db")

#Create a database cursor to interact with the database

cursor = conn.cursor()

#Create the table JOB

cursor.execute("""CREATE TABLE JOBS (JOB_CODE CHAR(3) PRIMARY KEY NOT NULL, 
               JOB_DESCRIPTION VARCHAR(15), JOB_CHG_HOUR DECIMAL(6,2), 
               JOB_LAST_UPDATE DATE);""")

#Insert a record into the table JOB

cursor.execute("INSERT INTO JOBS VALUES ('500', 'Programmer', 35.75, '2009-11-09')")

#Save the changes to database

conn.commit()

#Insert more records into the table JOB

cursor.execute("""INSERT INTO JOBS VALUES ('501', 'Systems Analyst', 96.75,  '2009-11-09'),
('502', 'Database Designer', 125.00, '2010-03-24'),
('503', 'Electrical Engineer', 84.50, '2009-11-09'),
('504', 'Mechanical Engineer', 67.90, '2009-11-09'),
('505', 'Civil Engineer', 55.78, '2009-11-09'),
('506', 'Clerical Support', 26.87, '2009-11-09'),
('507', 'DSS Analyst', 45.95, '2009-11-09'),
('508', 'Applications Designer', 48.10, '2010-03-24'),
('509', 'Bio Technician', 34.55, '2009-11-09'),
('510', 'General Support', 18.36, '2009-11-09');""")

conn.commit()

#Retrieve all the records in the table JOB

cursor.execute('SELECT * FROM JOBS')
jobs = cursor.fetchall()

for row in jobs:
    print(row)
```
